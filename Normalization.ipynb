```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('/content/wine_data.csv')
df.head()
```

```python
df = df.iloc[:,0:3]
df
```

```python
X = df[['alcohol','malic_acid']]
y = df['class_label']
```

```python
X_train,X_test,y_train,y_test= train_test_split(X,y,test_size = 0.3,random_state = 42)
```

```python
scaler = MinMaxScaler()

scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)
X_test_scaled= scaler.transform(X_test)
```

```python
np.round(X_train.describe(), 1)
```

```python
np.round(X_train_scaled.describe(), 1)
```

```python
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=X_test.columns)
```

```python
fig, (plot1, plot2) = plt.subplots(ncols=2, figsize=(12, 5))

plot1.scatter(X_train['alcohol'], X_train['malic_acid'],c=y_train)
plot1.set_title("Before Scaling")
plot2.scatter(X_train_scaled['alcohol'], X_train_scaled['malic_acid'],c=y_train)
plot2.set_title("After Scaling")
plt.show()
```

```python
fig, (plot1, plot2) = plt.subplots(ncols=2, figsize=(12, 5))

# before scaling
plot1.set_title('Before Scaling')
sns.kdeplot(X_train['alcohol'], ax=plot1)
sns.kdeplot(X_train['malic_acid'], ax=plot1)

# after scaling
plot2.set_title('After Standard Scaling')
sns.kdeplot(X_train_scaled['alcohol'], ax=plot2)
sns.kdeplot(X_train_scaled['malic_acid'], ax=plot2)
plt.show()
```

```python
fig, (plot1, plot2) = plt.subplots(ncols=2, figsize=(12, 5))

# before scaling
plot1.set_title('Alcohol Distribution Before Scaling')
sns.kdeplot(X_train['alcohol'], ax=plot1)

# after scaling
plot2.set_title('Alcohol Distribution After Standard Scaling')
sns.kdeplot(X_train_scaled['alcohol'], ax=plot2)
plt.show()
```

```python
fig, (plot1, plot2) = plt.subplots(ncols=2, figsize=(12, 5))

# before scaling
plot1.set_title('Malic Acid Distribution Before Scaling')
sns.kdeplot(X_train['malic_acid'], ax=plot1)

# after scaling
plot2.set_title('Malic Acid Distribution After Standard Scaling')
sns.kdeplot(X_train_scaled['malic_acid'], ax=plot2)
plt.show()
```
